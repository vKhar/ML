# Artificial Intelligence

There are many ways to define what an Artificial Intelligence (AI) system is. A very naive definition of AI is simply **a system capable of mimicking human intelligence**.

The definition used in the Cambridge O Level textbook is:

>The ability of a computer to perform
<u>complex tasks without constant human
guidance</u> and improve its performance
as more data is collected.

This definition tells us how an AI system behaves and adapts, but not what it is or how it is built. Secondly, it is **describing** the **behavior** of **a** specific category of AI system rather than providing a general definition of AI.

There is currently no universally accepted definition of what constitutes an AI system, the concept of AI remains an evolving one, shaped by advances in technology and the expanding and increasingly diverse applications of AI.

## Use cases for Machine Learning based AI System

Generally, Machine Learning AI is good at tasks where:
- The rules and goals are hard to write down, but examples are easy to find. No one can write a precise rule for what a handwritten ‘7’ looks like, but there are plenty of handwritten ‘7’ examples to learn from.
- The problem can be expressed through a measurable objective or performance criterion.
- Performance can improve as more relevant, representative training data are collected.
- Occasional errors are tolerable or recoverable, or a human can review and validate the output 

A task is usually a poor fit when:
- Data is scarce, unreliable or biased.
- The rules are simple and well-defined. Traditional programming is generally cheaper, faster to deploy, more predictable and easier to explain (e.g., calculating GST on a bill).
- Every decision must be fully explainable, or a single error would be severe and cannot be caught by a human.

| Category | Typical tasks and examples | Problem type | Why AI suits it | Main limitations and risks |
|---|---|---|---|---|
| Perception (images, video, audio) | Face detection, verification (e.g., phone unlock, Singpass) ; speech-to-text and speaker identification; image classification and object detection (medical scans, factory inspection, OCR, autonomous vehicles) | Classification; sequence processing | Rules for what a tumour or a cat looks like cannot be written by hand; deep neural networks learn features directly from labelled examples | Accents, noise, and under-represented groups reduce accuracy; identification raises privacy and ethical concerns |
| Language | Spam filtering, sentiment analysis, content moderation, machine translation, information extraction, summarisation, question answering, chatbots (LLMs) | Text classification and generation | Huge volumes of text; fixed rules are easily evaded; retraining on new data keeps up with changing tactics (e.g., spammers) | Adversarial users trying to defeat the system; legitimate content wrongly blocked; bias in training text |
| Structure discovery and anomaly detection | Customer segmentation and grouping related articles (clustering); intrusion detection, equipment faults, unusual transactions (anomaly detection) | Unsupervised learning | No labels needed; learns what normal looks like and flags deviations, which suits rare and varied problems | Clusters are unnamed and need human interpretation; rare anomalies are hard to evaluate |
| Generative AI | Drafting, editing and translating text; writing and explaining code; creating images, audio and video; captions and text-to-speech; synthetic data; AI agents that call tools | Generative modelling | Breadth and speed, especially for first drafts and brainstorming | Output is plausible, not verified: hallucinations, bias, copyright issues, deepfakes; agents have more autonomy and need more human oversight |

## Real World use case of a hybrid ML and GOFAI system

- Autonomous driving
    - ML handles tasks that are difficult to specify through explicit rules, such as object recognition, perception and behaviour prediction.
    - GOFAI handles tasks that benefit from explicitly defined rules, constraints, logical reasoning and deterministic decision-making.
- Modern Gaming engine
    - AlphaGo
        - uses Neural Network to learned from past games
        - uses Reinforcment learning to play against itself
        - uses Monte Carlo Tree Search (MCTS) to explored possible future moves to select promising strategies
