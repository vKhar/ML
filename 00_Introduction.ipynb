### Machine Learning Taxonomy of Learning Techniques


 There are many different training/learning techniques (about 20-30) in Machine Learning used to generate the AI Model, they can be classified as :

```mermaid
---
title: Machine Learning Taxonomy of Learning Techniques
---
flowchart TD
    A[Machine Learning]
    A --> B[<u>**Supervised**</u>]
    A --> C[<u>**Unsupervised**</u>]
    A --> F[Self-supervised]
    A --> D[Reinforcement]
    
```
- In **supervised learning**, the training data is labelled (i.e., the correct target output is provided for each data point). 
    - For instance, if the task were to classify songs into different genres of music, then the correct genre would already be provided for each song in the training data.
- In **unsupervised learning**, the training data does not need to be labelled. The output is then based on patterns in the data that are discovered by the algorithm. 
    - For instance, a simple unsupervised learning algorithm may classify data by grouping them into clusters based on similarity. 
    Unsupervised learning is useful if correct labels are not defined or readily available. 
    However, the output may be harder to interpret because the resulting classification groups are unnamed and not determined ahead of time.



### Machine Learning Terminology

| Term | Definition | Example |
|---|---|---|
| Data Set | The collection of data used for training and evaluation. | For the task of classifying songs into genres, the data set is a collection of songs, each described by its attributes and genre.  |
| Data point | Unit of information representing an instance or example relevant to the problem, a record, $d = (a_1, a_2, \dots, a_n)$ of attribute values. Typically, a row in the data set. | For the task of classifying songs into genres, each data point would represent a song.  |
| Feature | A feature is a measurable property of a data point used as an input, i.e. one of the attributes other than the label. | For the task of classifying songs into genres, possible features would be song length and song tempo. |
| Feature vector | The values of a data point's features, is often represented as a vector of m dimensions, written $\mathbf{x} = (x_1, \dots, x_m)$, where $x_j$ is the value of the $j$-th feature and $m \le n$ | A song with length 210 s and tempo 120 bpm: $\mathbf{x} = (210, 120)$. |
| Label | The target output the model should predict for a data point. In supervised learning, it is one of the data point's attributes and it cannot be a feature.| For the task of classifying songs into genres, the label would be the genre of each song: Rock, Jazz, Classical, etc. |
 Classification | A supervised learning task where the label is a **category** (a discrete class). The model assigns each data point to one of the possible classes. | Predicting a song's genre from its length and tempo. |
| Regression | A supervised learning task where the label is a **numerical value** (continuous). The model estimates a quantity for each data point. | Predicting a song's average loudness (dB) from its length and tempo. To train this model, the data point must now contain the loudness attribute as a label |


### ML vs Traditional Programming

| Aspect | Machine Learning | Traditional Programming |
|---|---|---|
| Approach | Providing a large amount of data for the system to identify patterns and implicitly formulate its own rules. | Writing explicit rules and instructions for every possible scenario. |
| Flexibility | Adapts to new data and can improve over time with more data. | Requires manual modification to change behaviour. |
| Predictability | Depending on the type of ML system, output may be probabilistic and vary each time for the same input. | Output is typically deterministic and repeatable for the same input. |
| Speed | Training speed is typically slow; inference speed varies greatly depending on the model's size and available resources. | Development and execution speeds are typically faster for well-defined problems with clear rules and requirements. |
| Cost | Typically higher cost due to need for specialised hardware (e.g., GPUs) and high power usage. | Typically lower cost due to lower hardware and power requirements. |
| Ideal use case | Solving problems with complex patterns that are difficult or impossible for humans to formulate explicit rules. | Solving well-defined problems with clear rules and requirements. |

### Python Machine Learning Libraries/Modules 

Machine Learning algorithms can, in principle, be implemented using pure Python code. However, Python has a rich ecosystem of libraries(modules) that provide ready-made implementations of most commonly used ML algorithms, together with supporting tools for building machine learning pipelines.

These libraries also leverage optimised, compiled code to accelerate computationally intensive operations, making them considerably more efficient than equivalent implementations written entirely in pure Python.
